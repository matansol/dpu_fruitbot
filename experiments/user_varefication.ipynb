```python
qualtrics_csv = "raw_data/qualtrics2_values.csv"
prolific_csv = "raw_data/prolific2_real.csv"
qualtrics_df = pd.read_csv(qualtrics_csv)

exp_start_time = pd.to_datetime("2026-03-25 00:00:00")
exp_end_time = pd.to_datetime("2026-03-31 23:00:00")


qualtrics_ids = qualtrics_df['ProlificID'].tolist()
qualtrics_df['StartDate'] = pd.to_datetime(qualtrics_df['StartDate'], errors='coerce')
qualtrics_df['EndDate'] = pd.to_datetime(qualtrics_df['EndDate'], errors='coerce')
# Filter the data
rellevant_qualtrics = qualtrics_df[
    (qualtrics_df['StartDate'] >= exp_start_time) & (qualtrics_df['StartDate'] <= exp_end_time) &
    (qualtrics_df['ProlificID'].astype(str).str.len() > 20) &
    ((qualtrics_df['Finished'] == 'True') | (qualtrics_df['Finished'] == '1'))  # Ensure the survey was completed
]
display(rellevant_qualtrics.tail(3))
```

```python
# prolific_csv = "raw_data/prolific_real1.csv"
prolific_df = pd.read_csv(prolific_csv)
prolific_df = prolific_df[prolific_df['Status'] == "AWAITING REVIEW"]
prolific_ids = prolific_df['Participant id'].tolist()
feedback_actions = all_tabels['feedback_actions']
feedback_user_ids = feedback_actions['user_id'].tolist()
actives_users = []
non_actives_users = []
for prolific_id in prolific_ids:
    # First check if user has bonus_group in rellevant_qualtrics
    user_qualtrics = rellevant_qualtrics[rellevant_qualtrics['ProlificID'] == prolific_id]
    if user_qualtrics.empty  or user_qualtrics['Finished'].iloc[0] == 'False': # or user_qualtrics['bonus_group'].isna().all()
        non_actives_users.append(prolific_id)
        continue

    if prolific_id not in feedback_user_ids:
        non_actives_users.append(prolific_id)
        continue
    
    actives_users.append(prolific_id)
save_path = "prolific_approves/approved_users.csv"
pd.DataFrame(actives_users, columns=['ProlificID']).to_csv(save_path, index=False)
print(f"Active users count: {len(actives_users)}")
print(f"non active users: {non_actives_users}")


# calculate the bonuses
# correct_answers = answer_dict
filtered_df_no_null = rellevant_qualtrics[rellevant_qualtrics['ProlificID'].isin(actives_users)]
bonus_fields = ["ProlificID", "Bonus"]

bonus_map = {"APPL1": 0, "APPL2": 0, "APPL3": 0.5, "APPL4": 0.5, "APPL5": 0.7, "APPL6": 1, 
             "2321": 0, "2322": 0.5, "2323": 1}
# filtered_df_no_null["Bonus"] = filtered_df_no_null["bonus_group"].map(bonus_map).fillna(0)
filtered_df_no_null["Bonus"] = filtered_df_no_null["Q171"].map(bonus_map).fillna(0)
"""bonus"""
bonus_df = filtered_df_no_null[filtered_df_no_null["Bonus"] != 0]
# df = get_bonus(df, bonus, correct_answers)
bonus_df = filtered_df_no_null[bonus_fields]
bonus_df = bonus_df[bonus_df["Bonus"] > 0]
bonus_df.to_csv("prolific_approves/bonus.csv", index=False)

```
